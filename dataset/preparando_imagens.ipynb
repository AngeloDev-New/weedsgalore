```python
import os
import cv2
import matplotlib.pyplot as plt
path_images = '../imagens'
# image_paths,image_name = 

for image_path,image_name in [(os.path.join(path_images,image_name),image_name) for image_name in os.listdir(path_images)]:
    image = cv2.imread(image_path,cv2.IMREAD_COLOR_RGB)
    r,g,b = cv2.split(image)
    cv2.imwrite(f'images/{image_name.replace('image','r_image')}',r)
    cv2.imwrite(f'images/{image_name.replace('image','g_image')}',g)
    cv2.imwrite(f'images/{image_name.replace('image','b_image')}',b)
```
