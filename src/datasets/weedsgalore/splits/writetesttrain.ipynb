```python
import os
train = 'train.txt'
test = 'test.txt'
dataset = '../../../../data/images'
with open(train,'w') as train:
    with open(test,'w') as test:
        # print(os.listdir('../../../../data/images'))  
        datas =  [img.replace('.png','\n') for img in os.listdir('../../../../data/images') if img.endswith('png')]
        datalen = len(datas)
        tamanhoTeste = int(datalen*0.2)
        # tamanhoTreino = datalen - tamanhoTeste
        for i,data in enumerate(datas):
            if i <  tamanhoTeste:
                test.write(data)
            else:
                train.write(data)
```
